```python
import matplotlib.pyplot as plt
import numpy as np

# Create a smooth polarization curve
current_density = np.linspace(0, 2.0, 300)  # A/cm^2
voltage = 1.23 - 0.15 * np.log(current_density + 0.02) - 0.3 * current_density - 0.05 * current_density**2

# Plot the curve only with axis labels
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(current_density, voltage, color='blue', linewidth=2)

# Axis labels only
ax.set_xlabel('Current Density (A/cm²)', fontsize=14)
ax.set_ylabel('Cell Voltage (V)', fontsize=14)

# Clean look: no grid, no ticks on top/right, no title or extra text
ax.tick_params(axis='both', labelsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Adjusted polarization curve to be more realistic:
# - Max voltage is now 1.2 V
# - End of curve includes a steeper nonlinear drop to reflect mass transport limitations

current_density = np.linspace(0, 2.0, 300)

# Define a more realistic polarization curve with three main regions
activation_loss = 0.08 * np.log(current_density + 0.02)
ohmic_loss = 0.25 * current_density
concentration_loss = 0.2 * (current_density**2)

voltage = 1.2 - activation_loss - ohmic_loss - concentration_loss

# Plot the refined curve
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(current_density, voltage, color='blue', linewidth=2)

# Axis labels only
ax.set_xlabel('Current Density (A/cm²)', fontsize=14)
ax.set_ylabel('Cell Voltage (V)', fontsize=14)

# Clean appearance
ax.tick_params(axis='both', labelsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Recreate a more realistic polarization curve
# Region 1: near-constant voltage with slow drop (activation)
# Region 2: linear drop (ohmic)
# Region 3: sharp curve down (mass transport)

current_density = np.linspace(0.01, 2.0, 300)

# Construct 3-part voltage loss manually
voltage = (
    1.2                                          # ideal starting voltage
    - 0.05 * np.log(current_density + 0.02)      # activation loss
    - 0.25 * current_density                     # ohmic loss
    - 0.04 * (current_density**3)                # concentration loss (more subtle)
)

# Plot the realistic polarization curve
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(current_density, voltage, color='blue', linewidth=2)

# Axis labels only
ax.set_xlabel('Current Density (A/cm²)', fontsize=14)
ax.set_ylabel('Cell Voltage (V)', fontsize=14)

# Clean appearance
ax.tick_params(axis='both', labelsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

```
